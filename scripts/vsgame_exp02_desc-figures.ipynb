### Import Packages

```python
import matplotlib.pyplot as plt
import glob
import os
import pandas as pd
import re
import seaborn as sns
import itertools
import ptitprince as pt
```

### Load Data

```python
base_dir = '/Users/aidelarazan/Box Sync/aidelarazan_box/Projects/vsgame/exp02/data/'
mst = pd.read_csv(os.path.join(base_dir, 'vsgame_exp02_sub-all_task-MST.csv'))
spatial_placement = pd.read_csv(os.path.join(base_dir, 'vsgame_exp02_sub-all_task-SpatialPlacement_desc-Accuracy.csv'))
item_desc = pd.read_csv(os.path.join(base_dir, 'vsgame_exp02_sub-all_task-ItemRecallDescriptions_desc-SemanticSimilarity.csv'))
spatial_desc = pd.read_csv(os.path.join(base_dir, 'vsgame_exp02_sub-all_task-SpatialRecallDescriptions_desc-SemanticSimilarity.csv'))
```

## MST

### dprime

```python
mst.head()
```

```python
fig, ax = plt.subplots(figsize=(5, 6), facecolor='none')

xtick_order = ['Active', 'Passive']
xtick_labels = {
    'Active': 'Active',
    'Passive': 'Passive',
}
xtick_colors = {
    'Active': '#006429',
    'Passive': '#80B294',    
}

# Swarmplot
pt.half_violinplot(
    data=mst,
    x='condition',
    y='dprime',
    hue='condition',
    hue_order=xtick_order,
    palette=xtick_colors, 
    bw=0.5, 
    cut=1, 
    scale="area", 
    width=0.5, 
    inner=None,
    order=xtick_order, 
    split=True,
    linewidth=0,
)

# Pointplot
sns.boxplot(
    data=mst,
    x='condition',
    y='dprime',
    palette=xtick_colors, 
    width=0.1, 
    order=xtick_order,
    fliersize=4,
    zorder=10,
    showcaps=False, 
    linewidth=2,
    boxprops={'facecolor': 'none', "zorder": 10, 'linewidth': 2},
    showfliers=True, 
    whiskerprops={'linewidth': 2, "zorder": 10}
)

# Pointplot
sns.stripplot(
    data=mst,
    x='condition',
    y='dprime',
    palette=xtick_colors, 
    edgecolor="black",
    size=7, 
    jitter=1, 
    zorder=0,
    order=xtick_order
)

# Axis labels and ticks
ax.set_ylabel("Basic Object Recognition\n(d')", fontsize=20)
ax.set_xlabel('')
ax.set_xticks(range(len(xtick_order)))
ax.set_xticklabels([xtick_labels[label] for label in xtick_order])

# Add significance line
xy = (0, 3.5)
xytext = (1, 3.5)
center = ((xy[0] + xytext[0]) / 2, (xy[1] + xytext[1]) / 2)

ax.annotate('', xy=xy, xytext=xytext, arrowprops=dict(arrowstyle='-', color='black'))
ax.text(center[0], center[1] + .1, '***', ha='center', va='bottom', fontsize=16)

# Adjust plot limits and aesthetics
ax.set_ylim(-2.1, 4)
ax.tick_params(labelsize=16)
ax.legend_.remove()

plt.savefig('/Users/aidelarazan/Box Sync/aidelarazan_box/Projects/vsgame/figures/vsgame_exp02_sub-all_task-MST_desc-dprime.png', dpi=300, bbox_inches='tight')
plt.show()
```
